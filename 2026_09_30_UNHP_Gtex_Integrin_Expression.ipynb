```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
integrins = pd.read_excel(r"gtex_integrin_7_organs.xlsx")
integrins
```

```python
brain_integrins = integrins[integrins['primary_site'] == 'Brain']
brain_integrins
```

```python
brain_genes_melted = brain_integrins.melt(
    id_vars=['Unnamed: 0', 'primary_site'],
    var_name='Integrin Gene',
    value_name='Expression Level'
)

brain_genes_melted['Expression Level'] = pd.to_numeric(brain_genes_melted['Expression Level'], errors='coerce')
brain_genes_melted.dropna(subset=['Expression Level'], inplace=True)

plt.figure(figsize=(15, 8))
sns.violinplot(x='Integrin Gene', y='Expression Level', data=brain_genes_melted, inner='quartile')
plt.title('Violin Plot of All Integrin Gene Expression Levels in Brain Tissue')
plt.xlabel('Integrin Gene')
plt.ylabel('Expression Level')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
liver_integrins = integrins[integrins['primary_site'] == 'Liver']
liver_integrins
```

```python
liver_genes_melted = liver_integrins.melt(
    id_vars=['Unnamed: 0', 'primary_site'],
    var_name='Integrin Gene',
    value_name='Expression Level'
)

liver_genes_melted['Expression Level'] = pd.to_numeric(liver_genes_melted['Expression Level'], errors='coerce')
liver_genes_melted.dropna(subset=['Expression Level'], inplace=True)

plt.figure(figsize=(15, 8))
sns.violinplot(x='Integrin Gene', y='Expression Level', data=liver_genes_melted, inner='quartile')
plt.title('Violin Plot of All Integrin Gene Expression Levels in Liver Tissue')
plt.xlabel('Integrin Gene')
plt.ylabel('Expression Level')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
brain_liver_integrins = integrins[integrins['primary_site'].isin(['Brain', 'Liver'])]

# Get the list of gene columns by dropping 'Unnamed: 0' and 'primary_site'
gene_columns = [col for col in integrins.columns if col not in ['Unnamed: 0', 'primary_site']]

brain_liver_integrins_vertical = brain_liver_integrins.melt(
    id_vars=['primary_site'],
    value_vars=gene_columns, # Explicitly specify gene columns to melt
    var_name='integrin_gene',
    value_name='expression_levels'
)
brain_liver_integrins_vertical
```

```python
brain_liver_integrins_vertical['expression_levels'] = pd.to_numeric(brain_liver_integrins_vertical['expression_levels'], errors='coerce')
brain_liver_integrins_vertical.dropna(subset=['expression_levels'], inplace=True)

plt.figure(figsize=(18, 9))
sns.violinplot(x='integrin_gene', y='expression_levels', hue='primary_site', data=brain_liver_integrins_vertical, inner='quartile', palette={'Brain': 'blue', 'Liver': 'orange'})
plt.title('Comparison of Integrin Gene Expression Levels in Brain vs. Liver Tissue')
plt.xlabel('Integrin Gene')
plt.ylabel('Expression Level')
plt.xticks(rotation=90)
plt.legend(title='Tissue Type')
plt.tight_layout()
plt.show()
```

```python
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

integrins = integrins.drop('Unnamed: 0', axis=1, errors='ignore')
integrins
```

```python
# Separate the gene expression data from the primary_site column
gene_expression_data = integrins.drop(columns=['primary_site'])

# Handle any potential non-numeric values or NaNs in the gene expression data
# Convert all columns to numeric, coercing errors to NaN
gene_expression_data = gene_expression_data.apply(pd.to_numeric, errors='coerce')
# Drop rows with any NaN values that might have resulted from the conversion
gene_expression_data.dropna(inplace=True)

# It's important to align the primary_site column with the cleaned gene_expression_data
# So, filter integrins to match the indices of gene_expression_data
integrins_filtered = integrins.loc[gene_expression_data.index]
primary_site_labels = integrins_filtered['primary_site']

# Standardize the data
scaler = StandardScaler()
scaled_gene_expression = scaler.fit_transform(gene_expression_data)

# Apply t-SNE to reduce dimensionality to 2D
tsne = TSNE(n_components=2, random_state=42)
tsne_results = tsne.fit_transform(scaled_gene_expression)

# Create a DataFrame for plotting
tsne_df = pd.DataFrame(data=tsne_results, columns=['TSNE_Component_1', 'TSNE_Component_2'])
tsne_df['Primary Site'] = primary_site_labels.values

# Visualize the t-SNE results
plt.figure(figsize=(12, 10))
sns.scatterplot(
    x='TSNE_Component_1',
    y='TSNE_Component_2',
    hue='Primary Site',
    data=tsne_df,
    palette='viridis',
    alpha=0.7
)
plt.title('t-SNE Plot of Integrin Gene Expression by Primary Site')
plt.xlabel('TSNE Component 1')
plt.ylabel('TSNE Component 2')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```
