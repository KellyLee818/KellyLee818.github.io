```python
import matplotlib.pyplot as plt
import numpy as np

# The line `ax.plot(x, x**2)` is used to plot data on the axes object `ax`.
# Specifically, it plots the values in the `x` array against their squares (`x**2`),
# creating a graph of the function y = x^2.
```

```python
x = np.linspace (0,10,100)
fig, ax = plt.subplots()
ax.plot(x,x**2)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("y=x^2")
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y1 = x**2
y2 = x**3

fig, ax = plt.subplots()
ax.plot(x, y1, label='y = x^2', color='blue')
ax.plot(x, y2, label='y = x^3', color='red')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Graph of y=x^2 and y=x^3')
ax.legend()
plt.grid(True)
plt.show()
```
