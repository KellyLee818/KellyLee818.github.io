```python
import pandas as pd
```

```python
chrom_data=pd.read_excel('/content/chrosome_basepairs.xlsx')
```

```python
chrom_data
```

```python
total_basepairs = chrom_data['basepais(bp)'].sum()
print(f"Total basepairs in the human genome: {total_basepairs}")
```

```python
total_basepairs
```

```python
angstrom_per_basepair = 3.4
total_length_angstrom = total_basepairs * angstrom_per_basepair
print(f"Total length of all basepairs stretched out: {total_length_angstrom} Angstroms")
```

```python
print(f"Total length of all basepairs stretched out:{total_length_angstrom} Angstroms")
```

```python
meters_per_angstrom = 1e-10
total_length_meters = total_length_angstrom * meters_per_angstrom
print(f"Total length of all basepairs stretched out: {total_length_meters} meters")
```

```python
print(f"Total length of all base pairs stretched out: {total_length_meters} meters")
```

```python
# Convert total basepairs to Megabasepairs (Mbps)
total_mbps = total_basepairs / 1_000_000
print(f"Total genome size: {total_mbps:.2f} Mbps")
```

```python
# Define costs per Mbps for given years
cost_per_mbps = {
    2001: 10000,   # $10,000 per Mbps
    2010: 1,       # $1 per Mbps
    2011: 0.10,    # $0.10 per Mbps
    2020: 0.01     # $0.01 per Mbps
}

# Calculate total cost for each year
costs = {}
for year, price_per_mbps in cost_per_mbps.items():
    costs[year] = total_mbps * price_per_mbps

# Display the calculated costs
for year, cost in costs.items():
    print(f"Total sequencing cost in {year}: ${cost:,.2f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a DataFrame for plotting
cost_df = pd.DataFrame(list(costs.items()), columns=['Year', 'Total_Cost'])

# Plot the total cost changes over the years
fig = plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='Total_Cost', data=cost_df, marker='o', linestyle='-')
plt.title('Human Genome Sequencing Cost Over Years')
plt.xlabel('Year')
plt.ylabel('Total Cost ($)')
plt.yscale('log') # Use a logarithmic scale due to large difference in costs
plt.grid(True, which="both", ls="--", c='0.7')
plt.xticks(list(cost_per_mbps.keys())) # Ensure all years are shown as ticks
plt.tight_layout()
plt.show()
```
